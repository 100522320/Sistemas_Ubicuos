### 1.4.Tipo de las variables:
Lo primero que haremos será ver con pandas y su función dtypes el tipo de cada variable. Asi podremos separarlas en numéricas o no numericas y este último grupo se separará posteriormente en categóricas y ordinales.

```python
print(df_datos.dtypes.values)
```

Como podemos ver todas son o int64 o Object por lo que podemos hacer un clasificador automático que las separe en numéricas (int64) o no numéricas (Object) que se dividirá en categóricas o ordinales. Para realizarlo haremos:

1. Dividir las variables en numéricas o no numéricas según su dtype utilizando la función select_dtypes.

2. De las no numéricas sacamos con unique los distintos valores que tienen y con nunique el numero total.

    *Esto nos servirá para ver que tanto education como month son variables ordinales y más tarde para sacar las categoricas con alta cardinalidad.

3. Creamos 2 listas más, una para el tipo (numéricas, categórica o ordinal) y otra para el rol (Target o Feature).

4. Recorrer el dataframe con un bucle y rellenando estas listas con los datos de las variables.

5. Crear un dataframe donde introduzcamos los datos de estas listas juntos al nombre y dtype de cada variable.

```python
# Clasificación automática de variables
variables_numericas = df_datos.select_dtypes(include=[np.number]).columns.tolist()
variables_no_numericas = df_datos.select_dtypes(include=['object']).columns.tolist()
target = 'deposit'

N_unique = []
Valores = []

for i in variables_no_numericas:
  N_unique.append(df_datos[i].nunique())
  Valores.append(df_datos[i].unique())

cardinalidad = pd.DataFrame({
    'Variable': variables_no_numericas,
    'N_unique': N_unique,
    'Valores': Valores
})
print(cardinalidad[['Variable', 'N_unique', 'Valores']].to_string(index=False))

variables_categoricas = []
variables_ordinales = []

for x in variables_no_numericas:
  if x in ('month','education'):
    variables_ordinales.append(x)
  else:
    variables_categoricas.append(x)


# Imprimimos la clasificación
print('\n=== Variables NUMÉRICAS ===')
print(variables_numericas)
print('\n=== Variables CATEGÓRICAS ===')
print(variables_categoricas)
print('\n=== Variables ORDINALES ===')
print(variables_ordinales)
print(f'\nTarget: {target}')

# Creamos listas vacías donde almacenar los tipos y roles de cada variable
lista_tipo = []
lista_rol = []

# Recorrer columnas
for c in df_datos.columns:
  # Tipo (Numérica, Categórica o Ordinal)
  if c in variables_numericas:
    lista_tipo.append('Numérica')
  elif c in variables_categoricas:
      lista_tipo.append('Categórica')
  else:
    lista_tipo.append('Ordinal')

  # Rol (Target o Feature)
  if c == target:
      lista_rol.append('Target')
  else:
      lista_rol.append('Feature')

# Crear DataFrame final
resumen_tipos = pd.DataFrame({
    'Variable': df_datos.columns,
    'Dtype': df_datos.dtypes.values,
    'Tipo': lista_tipo,
    'Rol': lista_rol
})

print('\n', resumen_tipos.to_string(index=True))
```

## 2.Decidir como se va a hacer la evaluación


### 2.1. Estimación del rendimiento futuro

Para la evaluación final del modelo (evaluación outer), utilizaremos la estrategia de Holdout. Dividiremos el dataset original en dos bloques: 66.6% para entrenamiento (2/3) y 33.3% para test (1/3).

Para la estimación del rendimiento futuro, utilizaremos la métrica de Accuracy. Dado que el ratio de nuestras clases está equilibrado (1.11), el Accuracy es una métrica representativa para medir el porcentaje de aciertos totales en la predicción de suscripciones al depósito


###2.2. División de los datos en Train y Test

Realizamos la partición utilizando la semilla de reproducibilidad (nuestro NIA) para garantizar que los experimentos sean comparables. El conjunto de test se reservará y solo se utilizará una vez al final de la práctica para la evaluación definitiva.

```python
from sklearn.model_selection import train_test_split

# Separamos variables de entrada (X) y objetivo (y)
X = df_datos.drop('deposit', axis=1)
y = df_datos['deposit']

# Partición Holdout: 2/3 entrenamiento, 1/3 test (test_size=0.33)
# Usamos stratify=y para que la proporción de 'sí/no' sea igual en ambos grupos
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.33,
    random_state=semilla,
    stratify=y
)

print(f"Instancias para entrenamiento y validación (Inner): {X_train.shape[0]}")
print(f"Instancias para estimación de rendimiento futuro (Outer): {X_test.shape[0]}")
```

### 2.3 Evaluación interna

Para la toma de decisiones durante la práctica (comparación de algoritmos KNN vs TREES y optimización de hiperparámetros - HPO), utilizaremos una evaluación interna (kf) basada en Validación Cruzada de KFold (K=3) aplicada exclusivamente sobre el conjunto de entrenamiento (X_train).

Se ha elegido un valor de K=3 para equilibrar la fiabilidad de la estimación con el coste computacional, especialmente considerando que los procesos de optimización de hiperparámetros (HPO) para modelos avanzados como SVM pueden ser intensivos en tiempo.

Este enfoque nos permite encontrar los mejores parámetros del modelo sin "contaminar" el conjunto de test, asegurando que la evaluación final sea honesta y libre de overfitting.


```python
from sklearn.model_selection import KFold

kf = KFold(n_splits=3, shuffle=True, random_state=semilla)
```

## 3.Metodos Básicos


### 3.1 Preprocesamiento


Para cumplir con el punto 3.1 del enunciado y evitar el Information Leakage, utilizaremos un Pipeline. Esto garantiza que el escalado se calcule solo en las particiones de entrenamiento de cada "fold" de la validación cruzada.

Las únicas variables que presentan valores faltantes en nuestro análisis son job y marital, ambas de naturaleza categórica. No obstante, siguiendo las buenas prácticas de preprocesamiento, hemos diseñado un Pipeline que incluye imputación para todos los tipos de variables: las numéricas se imputan mediante la mediana (para evitar la influencia de outliers) y las categóricas y ordinales mediante la moda (el valor más frecuente), garantizando así que el modelo sea robusto ante cualquier valor nulo presente o futuro. En el caso de las categóricas también usaremos OneHotEncoder (para transformar esas categorías en nuevas columnas con valores 0 o 1 si corresponden a ella) y para las ordinales usaremos OrdinalEncoder (que nos permite mantener en una sola columna el orden de las categorias pasandolas de str a numeros).

A continuacion se realizara la evaluación del mejor método de escalada usando KNN según su accuracy.

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, MinMaxScaler, RobustScaler, OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score

# Limpieza de seguridad: nos aseguramos de que el target no esté en las listas de variables
if 'deposit' in variables_categoricas:
    variables_categoricas.remove('deposit')

# Definimos el preprocesamiento para variables categóricas
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

# Definimos el orden lógico de nuestras variables ordinales.
educacion_orden = ['unknown', 'primary', 'secondary', 'tertiary']
meses_orden = ['jan', 'feb', 'mar', 'apr', 'may', 'jun', 'jul', 'aug', 'sep', 'oct', 'nov', 'dec']
categorias_ordenadas = [educacion_orden, meses_orden]

# Diccionario con los escaladores a testear
scalers = {
    'StandardScaler': StandardScaler(),
    'MinMaxScaler': MinMaxScaler(),
    'RobustScaler': RobustScaler()
}

results = {}

print("Comparando escaladores (Evaluación Inner - 3-fold CV):")

for name, scaler in scalers.items():
  # Definimos el preprocesamiento base para variables numéricas
  numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', scaler)
  ])

  # Definimos el preprocesamiento base para variables ordinales
  ordinal_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('ordinal', OrdinalEncoder(categories=categorias_ordenadas, handle_unknown='use_encoded_value', unknown_value=-1)),
    ('scaler', scaler)
  ])

  # Creamos el preprocesador completo uniendo numéricas, categóricas Y ordinales
  preprocessor = ColumnTransformer(
      transformers=[
          ('num', numeric_transformer, variables_numericas),
          ('cat', categorical_transformer, variables_categoricas),
          ('ord', ordinal_transformer, variables_ordinales)
      ])

  # Metemos todo en un Pipeline para evitar Data Leakage
  pipeline = Pipeline(steps=[
      ('preprocessor', preprocessor),
      ('knn', KNeighborsClassifier()) # Usamos k=5 por defecto
  ])

  # Ejecutamos Validación Cruzada sobre X_train
  scores = cross_val_score(pipeline, X_train, y_train, cv=kf, scoring='accuracy')

  results[name] = scores.mean()
  print(f"{name:15}: Accuracy medio = {scores.mean():.4f} ± {scores.std():.4f}")

# Seleccionamos el ganador
mejor_scaler_nombre = max(results, key=results.get)
print(f"RESULTADO: El mejor método de escalado es {mejor_scaler_nombre}")
```



---


**Le hemos añadido handle_unknown='use_encoded_value' y unknown_value=-1 por si en el conjunto de bank_competition_02.pkl apareciera algún valor raro que no esté en entrenamiento, para que el pipeline no falle.*


---

### 3.2 Evaluacion de hiperparametros

#### 3.2.1 Hiperparametros por omisión

A continuacion se evaluará el modelo usando KNN y arboles de decision mediante omision de hiperparametros y usando el mejor scaler de el apartado anterior

```python
import time
from sklearn.tree import DecisionTreeClassifier
# Usamos el scaler que salió como mejor método
prepro_ganador = ColumnTransformer(
    transformers=[
        ('num', Pipeline(steps=[
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', scalers[mejor_scaler_nombre])
        ]), variables_numericas),

        ('cat', categorical_transformer, variables_categoricas),

        ('ord', Pipeline(steps=[
            ('imputer', SimpleImputer(strategy='most_frequent')),
            ('ordinal', OrdinalEncoder(categories=categorias_ordenadas, handle_unknown='use_encoded_value', unknown_value=-1)),
            ('scaler', scalers[mejor_scaler_nombre])
          ]), variables_ordinales)
    ])

# Diccionario de modelos con parámetros por omisión
modelos_base = {
    'KNN (k_vecinos=5)': Pipeline(steps=[
        ('preprocessor', prepro_ganador),
        ('clf', KNeighborsClassifier()) # Por defecto n_neighbors=5
    ]),
    'Árbol de Decisión': Pipeline(steps=[
        ('preprocessor', prepro_ganador),
        ('clf', DecisionTreeClassifier(random_state=semilla))
    ])
}

print(f"Comparación con parámetros por defecto (Mejor Scaler: {mejor_scaler_nombre})")

for nombre, modelo in modelos_base.items():
    # Medimos el tiempo de entrenamiento (fit) sobre todo el X_train
    t_inicio = time.time()
    modelo.fit(X_train, y_train)
    t_entrenamiento = time.time() - t_inicio

    # Evaluación Inner para obtener el Accuracy medio
    scores = cross_val_score(modelo, X_train, y_train, cv=kf, scoring='accuracy')

    print(f"{nombre:18} | Accuracy CV: {scores.mean():.4f} ± {scores.std():.4f} | Tiempo Fit: {t_entrenamiento:.4f}s")
```

Si analizamos los resultados obtenidos podemos ver que tanto KNN como Árbol de Decisión tienen un accuracy muy similar. Sin embargo, el tiempo de ejecución es mucho menor en KNN haciendo que por el momento parezca la mejor opción.

#### 3.2.2 Interpretación de arboles poco profundos

Un árbol de decisión normal crecerá hasta que clasifique perfectamente, o casi, todos los datos de entrenamiento, lo que genera un árbol gigante. Si le limitamos la profundidad (por ejemplo, a 3 niveles), nos mostrará únicamente las decisiones más importantes que dividen a los clientes entre los que suscriben el depósito y los que no.

Para ver cuales son estas decisiones haremos un árbol con max_depth=3 y lo dibujaremos con plot_tree para que se interprete y entienda mejor.

```python
from sklearn.tree import plot_tree

# Creamos un árbol poco profundo (max_depth=3) usando nuestro pipeline ganador
arbol_interpretable = Pipeline(steps=[
    ('preprocessor', prepro_ganador),
    ('clf', DecisionTreeClassifier(max_depth=3, random_state=semilla))
])

# Entrenamos el modelo con los datos
arbol_interpretable.fit(X_train, y_train)

# Extraemos los nombres de las variables despues del preprocesamiento
nombres_caracteristicas = arbol_interpretable.named_steps['preprocessor'].get_feature_names_out()

# Extraemos las clases objetivo (['no', 'yes'])
clases_objetivo = arbol_interpretable.named_steps['clf'].classes_

# Dibujamos el árbol
plt.figure(figsize=(20, 10))
plot_tree(
    arbol_interpretable.named_steps['clf'],
    feature_names=nombres_caracteristicas,
    class_names=clases_objetivo,
    filled=True,
    rounded=True,
    fontsize=10,
    proportion=True # Muestra porcentajes en lugar de números absolutos para entenderlo mejor
)
plt.title("Árbol de Decisión Interpretable (Profundidad Máxima = 3)", fontsize=15)
plt.tight_layout()
plt.show()
```

**Colores:**

En este caso, los cuadrados serán más naranjas o más azules dependiendo de si es mas probable que se elija un no o un yes, respectivamente. Cuanto mayor sea la probabilidad de uno respecto al otro más oscuro se verá, por ello los casos en los que ambas probabilidades sean muy parecidas el cuadrado será blanco.

Para el HPO se utilizarán 50 iteraciones. La explicación de porque 50 y no más o menos se ira explicando a lo largo del apartado según se vayan estudiando diferentes graficas y datos.

Además, se tendrán en cuenta los hiperparámetros de profundidad máxima (max_depth), para limitar el crecimiento del árbol, mínimo de muestras por división (min_samples_split) y mínimo de muestras por hoja (min_samples_leaf), para controlar la granularidad de los nodos, el parámetro de post-podado ccp_alpha, que indica el umbral de complejidad de coste para eliminar ramas que no aportan una mejora significativa al modelo, reduciendo así el riesgo de sobreajuste. Finalmente, se evaluará el criterion, el cual determina la función métrica utilizada para medir la calidad de cada división en los nodos.

```python
import optuna
from optuna_integration import OptunaSearchCV
from optuna.distributions import IntDistribution as IntDist
from optuna.distributions import FloatDistribution as FloatDist
from optuna.distributions import CategoricalDistribution as CatDist
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Definimos el espacio de búsqueda
param_grid_optuna = {
  'clf__max_depth': IntDist(2, 30),
  'clf__min_samples_leaf': IntDist(2, 25),
  'clf__ccp_alpha': FloatDist(1e-5, 0.005, log=True),
  'clf__min_samples_split': IntDist(2, 20),
  'clf__criterion': CatDist(['gini', 'entropy'])
}

# Creamos el Pipeline que une el preprocesador con el modelo
pipeline_tree = Pipeline(steps=[
    ('preprocessor', prepro_ganador),
    ('clf', DecisionTreeClassifier(random_state=semilla))
])

# Configuración de OptunaSearchCV
budget = 50
tree_optuna = OptunaSearchCV(
    pipeline_tree,
    param_grid_optuna,
    scoring='accuracy',
    n_trials=budget,
    cv=kf,
    n_jobs=1,
    verbose=1,
    random_state=semilla
)

# Ejecución del HPO
# Guardamos los tiempos para la comparación
np.random.seed(semilla)
t_inicio_tree = time.time()

tree_optuna.fit(X_train, y_train)

t_final_tree = time.time()
duracion_tree = t_final_tree - t_inicio_tree

print("-" * 50)
print(f"Mejores parámetros Árbol de decisión: {tree_optuna.best_params_}")
print(f"Mejor accuracy: {tree_optuna.best_score_:.4f}")
```

```python
# Definimos el espacio de búsqueda
param_grid_svm = {
    'clf__C': FloatDist(1e-3, 1e3, log=True),
    'clf__gamma': FloatDist(1e-4, 1, log=True)
}

# Creamos el Pipeline que une el preprocesador con el modelo
pipeline_svm = Pipeline(steps=[
    ('preprocessor', prepro_ganador),
    ('clf', SVC(kernel='rbf',random_state=semilla))
])

# Configuración de OptunaSearchCV
optuna_svm = OptunaSearchCV(
    pipeline_svm,
    param_grid_svm,
    scoring='accuracy',
    n_trials=20, # Budget
    cv=kf,
    random_state=semilla
)

t_inicio = time.time()
optuna_svm.fit(X_train, y_train)
t_hpo_svm = time.time() - t_inicio

print("-" * 50)
print(f"Mejores Parametros: {optuna_svm.best_params_}")
print(f"Mejor Accuracy SVM: {optuna_svm.best_score_:.4f}")
```

###  5.2 Random Forest

Ahora probaremos con Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

# Definimos el espacio de búsqueda específico para Random Forest
param_grid_optuna = {
    'clf__n_estimators': IntDist(50, 300),           # Número de árboles en el bosque (equivalente a iterations)
    'clf__max_depth': IntDist(2, 30),                # Profundidad máxima de cada árbol
    'clf__min_samples_split': IntDist(2, 10),        # Mínimo de muestras para dividir un nodo
    'clf__min_samples_leaf': IntDist(1, 10),         # Mínimo de muestras que debe tener una hoja final
    'clf__criterion': CatDist(['gini', 'entropy'])   # Criterio para medir la calidad del split
}

# Creamos el Pipeline que une el preprocesador con el modelo
pipeline_rf = Pipeline(steps=[
    ('preprocessor', prepro_ganador),
    ('clf', RandomForestClassifier(random_state=semilla, n_jobs=1))
])

# Configuración de OptunaSearchCV
budget = 50
rf_optuna = OptunaSearchCV(
    pipeline_rf,
    param_grid_optuna,
    scoring='accuracy',
    n_trials=budget,
    cv=kf,
    n_jobs=1,
    verbose=1,
    random_state=semilla
)

# Ejecución del HPO
# Guardamos los tiempos para la comparación
np.random.seed(semilla)
t_inicio_rf = time.time()

rf_optuna.fit(X_train, y_train)

t_final_rf = time.time()
duracion_rf = t_final_rf - t_inicio_rf

print("-" * 50)
print(f"Mejores parámetros Random Forest: {rf_optuna.best_params_}")
print(f"Mejor accuracy: {rf_optuna.best_score_:.4f}")
print(f"Tiempo de ejecución: {duracion_rf:.2f} segundos")
```

Como podemos observar, Random Forest nos da un accuracy de 0,8399, un resultado sólido pero que se queda por debajo del 0,8594 alcanzado previamente con el modelo CatBoost. Esta diferencia de rendimiento tiene sentido desde el punto de vista teórico abordado en el proyecto. Mientras que Random Forest es un modelo basado en Bagging que construye múltiples árboles independientes para reducir la varianza y evitar el sobreajuste, CatBoost pertenece a la familia de Gradient Boosting. Este último construye los árboles de forma secuencial, enfocándose en corregir los errores residuales de las iteraciones anteriores, lo que frecuentemente le otorga un mayor techo de precisión en datos tabulares. Además, la naturaleza de nuestro dataset, compuesto por numerosas variables categóricas (algunas de alta cardinalidad), beneficia estructuralmente al enfoque de CatBoost.Debido a la evidente superioridad de catboost no vemos necesarios realizar un analisis en mayor profundidad de randomforest y pasaremos a probar otro modelo.

### 5.3 HistGradientBoosting

Probamos HistGradientBoosting

```python
from sklearn.ensemble import HistGradientBoostingClassifier

# Definimos el espacio de búsqueda para HistGradientBoosting
param_grid_optuna = {
    'clf__max_iter': IntDist(50, 300),
    'clf__learning_rate': FloatDist(0.01, 0.3, log=True),   # Tasa de aprendizaje
    'clf__max_depth': IntDist(2, 20),
    'clf__min_samples_leaf': IntDist(1, 30),
    'clf__l2_regularization': FloatDist(1e-3, 10.0, log=True) # Penalización L2 para estabilizar las hojas
}

# Creamos el Pipeline que une el preprocesador con el modelo
pipeline_hgb = Pipeline(steps=[
    ('preprocessor', prepro_ganador),
    ('clf', HistGradientBoostingClassifier(random_state=semilla))
])

# Configuración de OptunaSearchCV
budget = 50
hgb_optuna = OptunaSearchCV(
    pipeline_hgb,
    param_grid_optuna,
    scoring='accuracy',
    n_trials=budget,
    cv=kf,
    n_jobs=1,
    verbose=1,
    random_state=semilla
)

# Ejecución del HPO
# Guardamos los tiempos para la comparación
np.random.seed(semilla)
t_inicio_hgb = time.time()

hgb_optuna.fit(X_train, y_train)

t_final_hgb = time.time()
duracion_hgb = t_final_hgb - t_inicio_hgb

print("-" * 50)
print(f"Mejores parámetros HistGradientBoosting: {hgb_optuna.best_params_}")
print(f"Mejor accuracy: {hgb_optuna.best_score_:.4f}")
print(f"Tiempo de ejecución: {duracion_hgb:.2f} segundos")
```

El acuraccy de HistGradientBoosting es 0.8593 lo cual es exactamente igual al de catboosting de 0.8594 pero el primero tarda 138 segundos en entrenar con respecto a los 179 del segundo esto es porque esto es porque HistGradientBoosting utiliza internamente una técnica de discretización de los datos basada en la creación de histogramas (agrupando los valores continuos en bins o intervalos). En lugar de evaluar cada valor individual para encontrar el punto de corte óptimo en los nodos del árbol, evalúa únicamente los límites de estos histogramas, lo que reduce drásticamente el coste computacional y la memoria necesaria.

Tal y como señala la base teórica del proyecto, HistGradientBoosting es una "implementación más eficiente, basada en LightGBM, ideal para cuando se manejan muchos datos". Además, dado que en nuestro preprocesamiento final hemos transformado las variables categóricas en múltiples columnas numéricas mediante One-Hot Encoding, la técnica de histogramas de HistGradientBoosting consigue procesar esta extensa matriz de ceros y unos con mayor agilidad que el motor de CatBoost. De este modo, logramos el mismo poder predictivo y capacidad de generalización (Gradient Boosting), pero optimizando el tiempo de entrenamiento en un 23% menos.

Esto es coherente con la teoría de Gradient Boosting, ya que controlar la profundidad de los modelos base es una de las vías principales para gestionar el nivel de sobreajuste. Le siguen en relevancia la tasa de aprendizaje (learning_rate con un 23%) y el número de árboles (max_iter con un 17%), dos parámetros que operan de forma conjunta para estabilizar el modelo. Por último, los parámetros específicos de regularización en las hojas (min_samples_leaf y l2_regularization) tuvieron un impacto residual en el rendimiento final, sumando apenas un 5% de importancia conjunta.

### 6.2 Estimación del rendimiento futuro

Habiendo elegido el modelo final ahora tenemos que estimar como lo hará en la competición. Para esto usaremos los datos de test que dividimos en un inicio y dejamos de lado hasta este momento para que se mantuviesen intactos. A parte usaremos matrices de confusión, que nos darán los datos repartidos en un 2x2 que nos informará sobre los verdaderos positivos, verdaderos negativos, falsos positivos y falsos negativos. Esto nos dará más información sobre los fallos del modelo que el accuracy que es un simple porcentaje de aciertos.

```python
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

# Usamos el mejor modelo encontrado para predecir sobre el conjunto de test (Outer)
y_pred_test = hgb_optuna.predict(X_test)

# Métrica principal
accuracy_outer = accuracy_score(y_test, y_pred_test)
print(f"Accuracy en Test (Estimación Outer): {accuracy_outer:.4f}")

# Reporte completo
print("\nReporte de Clasificación:")
print(classification_report(y_test, y_pred_test))

# Matriz de Confusión
cm = confusion_matrix(y_test, y_pred_test)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=hgb_optuna.classes_)
disp.plot(cmap=plt.cm.Blues)
plt.title("Matriz de Confusión - Modelo Final (Outer)")
plt.show()
```

La evaluación en el conjunto de test (Outer) arroja un Accuracy del 87.13%, confirmando la robustez del modelo sin signos de sobreajuste respecto a la evaluación Inner. Analizando la matriz de confusión desde una perspectiva de negocio, el modelo es altamente efectivo: presenta un excelente Recall del 90% para la clase minoritaria ('yes'). Esto significa que el banco lograría contactar e identificar al 90% de los clientes predispuestos a suscribir el depósito, minimizando la pérdida de oportunidades (solo 171 falsos negativos), a cambio de un coste asumible en llamadas no fructíferas (296 falsos positivos).
